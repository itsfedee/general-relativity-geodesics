# 3. Kerr and Majumdar-Papapetrou

Kerr in Boyer-Lindquist coordinates (with a closed-form inverse metric) and the Majumdar-Papapetrou multi-black-hole solution. Both go through the same `solve_geodesic` / `animate_geodesic` code path.

```python
import numpy as np
import matplotlib.pyplot as plt
import plotly.graph_objects as go
import plotly.io as pio
pio.renderers.default = 'notebook'

from geodesics import Kerr, MajumdarPapapetrou, solve_geodesic, resample, animate_geodesic, plot_radial_velocity
```

## Kerr, $a = 0.9$

Yellow dots on the horizon rotate with $\Omega_H$; the translucent surface is the ergosphere. `resample(..., power=0.4)` redistributes the frames so the final plunge is not compressed into the last few.

```python
kerr = Kerr(M_val=1.0, a_val=0.9)
x0, v0 = kerr.from_cartesian(x=10, y=0, z=0, vx=-0.3, vy=0.0, vz=0.0, massive=True)
sol = solve_geodesic(kerr, x0, v0, (0, 250))
resample(sol, n=1000, power=0.4)
print(f"r_end = {sol.y[1][-1]:.3f}, r_+ = {kerr.r_horizon:.3f}, r_isco = {kerr.r_isco:.3f}")
```

```python
animate_geodesic(sol, kerr, n_frames=150, duration=20, coordinate_time=True).show()
```

```python
plot_radial_velocity(sol);
```

```python
plt.plot(sol.t, sol.y[1])
plt.axhline(kerr.r_horizon, color='k', ls='--', label='$r_+$')
plt.xlabel('τ'); plt.ylabel('r'); plt.legend(); plt.show()
```

### Equatorial embedding

The equatorial plane of Kerr embedded as a surface of revolution. Switching the view does not require re-integrating.

```python
kerr_emb = Kerr(M_val=1.0, a_val=0.6, embedding=True)
x0, v0 = kerr_emb.from_cartesian(x=20, y=0, z=0, vx=-0.3, vy=0, vz=0, massive=True)
sol = solve_geodesic(kerr_emb, x0, v0, (0, 200))
animate_geodesic(sol, kerr_emb, n_frames=70, duration=15).show()

kerr_emb.embedding = False
animate_geodesic(sol, kerr_emb, n_frames=70, duration=15).show()
```

## Majumdar-Papapetrou

Two extremal charged black holes in static equilibrium. The metric is written in isotropic Cartesian coordinates, so `from_cartesian` is the identity. Horizons sit at $r_i = 0$; the integration stops when the particle gets close to a centre.

```python
mp = MajumdarPapapetrou(masses=[1.0, 20.0], positions=[[0, 5, 0], [0, -5, 0]])
x0, v0 = mp.from_cartesian(x=10.0, y=3.0, z=0.0, vx=-0.5, vy=0.0, vz=0.0, massive=True)
sol = solve_geodesic(mp, x0, v0, (0, 50), t_eval=np.linspace(0, 50, 1000))
print(f"stopped at tau = {sol.t[-1]:.2f}")
animate_geodesic(sol, mp, n_frames=50, duration=15).show()
```
