# 1. Metrics and Christoffel symbols

A `Metric` is a sympy matrix plus its coordinates. From it we can print the line element, compute the Christoffel symbols and write down the geodesic equations symbolically.

```python
import sympy as sp
sp.init_printing()

from geodesics import (Sphere, Schwarzschild, Kerr, ReissnerNordstrom, MajumdarPapapetrou,
                       compute_christoffel, print_metric, print_christoffel, print_geodesic_eqs)
```

## Line elements

```python
S2 = Sphere(2)
print_metric(S2)
```

```python
schw = Schwarzschild(dim=4)
print_metric(schw)
```

```python
kerr = Kerr(M_val=1.0, a_val=0.9)
print_metric(kerr)
```

```python
kerr.g
```

## Christoffel symbols

$$\Gamma^\sigma_{\mu\nu} = \tfrac12 g^{\sigma\rho}\left(\partial_\mu g_{\nu\rho} + \partial_\nu g_{\mu\rho} - \partial_\rho g_{\mu\nu}\right)$$

Only the non-zero, upper-triangular ($\mu \le \nu$) entries are shown.

```python
_, nz_S2 = compute_christoffel(S2)
print_christoffel(S2, nz_S2)
```

```python
_, nz_schw = compute_christoffel(schw)
print_christoffel(schw, nz_schw)
```

## Geodesic equations

$$\ddot x^\sigma = -\Gamma^\sigma_{\mu\nu}\,\dot x^\mu \dot x^\nu$$

```python
print_geodesic_eqs(S2, nz_S2)
```

```python
print_geodesic_eqs(schw, nz_schw)
```

## A messy one: Majumdar-Papapetrou

For two extremal charged black holes the symbols are large rational functions of $(x,y,z)$. `simplify=False` uses `sp.cancel` instead of `sp.simplify`, which is much faster here. This is why the numerical integrator does **not** go through the symbolic Christoffels: it lambdifies the acceleration directly (see `geodesics.integrate`).

```python
import time
mp = MajumdarPapapetrou(masses=[1.0, 1.0], positions=[[0, 5, 0], [0, -5, 0]])
t0 = time.time()
_, nz_mp = compute_christoffel(mp, simplify=False)
print(f"{len(nz_mp)} non-zero symbols in {time.time()-t0:.1f}s")
for key, expr in list(nz_mp.items())[:5]:
    print(key, sp.count_ops(expr), "ops")
```
