# 2. Integrating geodesics: $S^2$ and Schwarzschild

The geodesic equation is rewritten as a first-order system
$\dot x^\mu = v^\mu,\; \dot v^\sigma = -\Gamma^\sigma_{\mu\nu} v^\mu v^\nu$
and integrated with DOP853. Initial conditions are given in Cartesian coordinates and converted by `metric.from_cartesian`, which also fixes $v^t$ from the normalisation $g_{\mu\nu}v^\mu v^\nu = -1$ (massive) or $0$ (light).

```python
import numpy as np
import plotly.io as pio
pio.renderers.default = 'notebook'

from geodesics import Sphere, Schwarzschild, solve_geodesic, animate_geodesic, plot_radial_velocity
```

## Great circle on $S^2$

```python
S2 = Sphere(2)
x0 = np.array([np.pi/2, 0.0])   # (theta, phi) at tau = 0
v0 = np.array([1.0, 0.0])
sol = solve_geodesic(S2, x0, v0, (0, 0.5*np.pi), t_eval=np.linspace(0, 0.5*np.pi, 1000))
print("theta range:", sol.y[0].min(), sol.y[0].max())
```

```python
animate_geodesic(sol, S2, n_frames=25).show()
```

## Infall into a Schwarzschild black hole

The integration stops automatically at the horizon (event index 0). With `coordinate_time=True` the animation is sampled uniformly in $t$, so the particle visibly freezes as it approaches $r = 2M$.

```python
schw = Schwarzschild(dim=4, M_val=0.75, embedding=True)
x0, v0 = schw.from_cartesian(x=10.0, y=4.0, z=0.0, vx=-0.5, vy=0.1, vz=0.0, massive=True)
sol = solve_geodesic(schw, x0, v0, (0, 80), t_eval=np.linspace(0, 80, 1000))

print(f"points: {len(sol.t)},  r_min = {sol.y[1].min():.3f},  horizon at {schw.r_horizon}")
print(f"t at 50% / 90% of the points: {sol.y[0][len(sol.t)//2]:.1f} / {sol.y[0][int(0.9*len(sol.t))]:.1f}")
```

```python
animate_geodesic(sol, schw, coordinate_time=True, n_frames=100).show()
```

Same solution, standard 3D view instead of the Flamm paraboloid:

```python
schw.embedding = False
animate_geodesic(sol, schw, coordinate_time=True, n_frames=100).show()
```

$|dr/d\tau|$ keeps growing, while $|dr/dt|$ goes to zero at the horizon:

```python
plot_radial_velocity(sol);
```
